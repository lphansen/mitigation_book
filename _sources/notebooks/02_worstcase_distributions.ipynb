
<!-- ——— Reproduction instructions ——— -->
<div style="margin:0.8rem 0 0.2rem;">
To reproduce Table 3 in the paper, run the shell scripts provided below.
</div>

~~~~bash
bash ./sbatch/postdamage.sh --model full_model
bash ./sbatch/postdamage_sub.sh --model full_model calibrating_xia 1
bash ./sbatch/predamage.sh --model full_model calibrating_xia 1
bash ./sbatch/feymannkacs_prepare.sh --model full_model 1
bash ./sbatch/postdamage_sub.sh --model full_model calibrating_xia 3
bash ./sbatch/predamage.sh --model full_model calibrating_xia 3
bash ./sbatch/feymannkacs_prepare.sh --model full_model 3

bash ./sbatch/postdamage.sh --model full_model --grid-profile double_resolution --initial-guess-grid-profile baseline
bash ./sbatch/predamage.sh --model full_model --grid-profile double_resolution --initial-guess-grid-profile baseline baseline_xia
bash ./sbatch/feymannkacs_prepare.sh --model full_model --grid-profile double_resolution baseline_xia

bash ./sbatch/update_results.sh --model full_model --grid-profile double_resolution --full-model-drift-only
~~~~

After the run finishes, the prepare job prints `h_k(X0) = ...` and `h_r(X0) = ...` to its job-out under `job-outs/<dataset>/feymannkacs_prepare/`.

Source scripts: 
* [**`postdamage.sh`**](https://github.com/korito1416/two-capital-climate-change/blob/main/sbatch/postdamage.sh)
* [**`postdamage_sub.sh`**](https://github.com/korito1416/two-capital-climate-change/blob/main/sbatch/postdamage_sub.sh)
* [**`predamage.sh`**](https://github.com/korito1416/two-capital-climate-change/blob/main/sbatch/predamage.sh)
* [**`feymannkacs_prepare.sh`**](https://github.com/korito1416/two-capital-climate-change/blob/main/sbatch/feymannkacs_prepare.sh)

## Distorted density for climate models <a href="https://github.com/korito1416/two-capital-climate-change/blob/main/python/zeroshock_trajectories_simulate.py" class="code-link" target="_blank" rel="noopener"></a>


The decision-maker has a baseline probability $dP_t(\theta)$ that weights each of the 144 climate models equally at every time instant, $t$, and makes an adjustment for ambiguity by solving Problem 3.2 of the paper:

$$
\min_{Q_t \ge 0,\; \int_\Theta Q_t(\theta)\,dP_t(\theta)=1}\;
\frac{\partial V}{\partial y}(X_t)\,\mathcal{E}_t \int_\Theta \theta\,Q_t(\theta)\,dP_t(\theta)
\;+\;
\xi_a \int_\Theta Q_t(\theta)\,\log Q_t(\theta)\,dP_t(\theta)
$$

with the minimizing relative density given by an exponential tilt:

$$
Q_t^{\ast}(\theta)
\;\propto\;
\exp\!\Bigl[-\frac{1}{\xi_a}\,\frac{\partial V}{\partial y}(X_t)\,\mathcal{E}_t\,\theta\Bigr].
$$

<!--
<div style="
     display:flex;
     gap:18px;
     justify-content:center;
     align-items:center;
     flex-wrap:wrap;
">
  <img src="../_static/20_damage/climate_xi005.png"
       alt="more aversion"
       style="width:350px; height:auto;">
  <img src="../_static/20_damage/climate_xi010.png"
       alt="less aversion"
       style="width:350px; height:auto;">
</div>
-->


<div style="
     display:flex;
     justify-content:center;
     align-items:center;
">
  <img src="../_static/paper/Figure4_climate_model_relative_weights_year-0.png"
       alt="Uncertainty-adjusted relative densities for climate models"
       style="width:700px; max-width:100%; height:auto;">
</div>

<!-- ——— Reproduction instructions ——— -->
<div style="margin:0.8rem 0 0.2rem;">
  To reproduce Figure 4 in the paper, run the shell scripts provided below.
</div>

```bash
bash ./sbatch/postdamage.sh --model full_model
bash ./sbatch/postdamage_sub.sh --model full_model calibrating_xia 1
bash ./sbatch/predamage.sh --model full_model calibrating_xia 1
bash ./sbatch/feymannkacs_prepare.sh --model full_model 1
bash ./sbatch/postdamage_sub.sh --model full_model calibrating_xia 3
bash ./sbatch/predamage.sh --model full_model calibrating_xia 3
bash ./sbatch/feymannkacs_prepare.sh --model full_model 3

bash ./sbatch/postdamage.sh --model full_model --grid-profile double_resolution --initial-guess-grid-profile baseline
bash ./sbatch/predamage.sh --model full_model --grid-profile double_resolution --initial-guess-grid-profile baseline baseline_xia
bash ./sbatch/zeroshock_trajectories_simulate.sh --grid-profile double_resolution true false false
bash ./sbatch/zeroshock_trajectories_plot.sh --grid-profile double_resolution

bash ./sbatch/update_results.sh --model full_model --grid-profile double_resolution
```

## Distorted density for the time of the first jump  <a href="https://github.com/korito1416/two-capital-climate-change/blob/main/python/feymannkacs_simulate.py" class="code-link" target="_blank" rel="noopener"></a>

By conditioning first on the state-variable history, we find that the probability that a jump has occurred by time $t$ under the uncertainty-adjusted probabilities is

$$
F_t \;=\; 1 - \exp\!\Bigl[ - \int_{0}^{t} \sum_{\ell=1}^{L}
        G^{*\ell}_\tau\,\mathcal J^{\ell}\!\bigl(X_\tau\bigr)\, d\tau \Bigr],
$$

which increases in $t$.
The implied density obtained by differentiating with respect to $t$ is

$$
f_t \;=\; \sum_{\ell=1}^{L} G^{*\ell}_t\,\mathcal J^{\ell}(X_t)\,
        \exp\!\Bigl[ - \int_{0}^{t} \sum_{\hat\ell=1}^{L}
        G^{*\hat\ell}_\tau\,\mathcal J^{\hat\ell}\!\bigl(X_\tau\bigr)\, d\tau \Bigr].
$$

We unpack this density by jump type,

$$
f_t \;=\; \sum_{\ell=1}^{L} f_t^{\ell},
$$

where

$$
f_t^{\ell} \overset{\mathrm{def}}{=} G^{*\ell}_t\,\mathcal J^{\ell}(X_t)\,
\exp\!\Bigl[ - \int_{0}^{t} \sum_{\hat\ell=1}^{L}
            G^{*\hat\ell}_\tau\,\mathcal J^{\hat\ell}\!\bigl(X_\tau\bigr)\, d\tau \Bigr],
\qquad \ell = 1,\dots,L.
$$
