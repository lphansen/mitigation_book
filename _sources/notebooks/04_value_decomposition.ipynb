# Value Decomposition

## Mathematical Formulation
$$
\frac{\partial V}{\partial x}(X_0)\cdot\Lambda_0
=
\widetilde{\mathbb E}
\left[
\int_0^\infty
Dis_t\left(\Lambda_t\cdot Scf_t\right)\,dt
\mid X_0,\Lambda_0
\right].
$$

where

$$
\begin{align*}
Dis_t
&\overset{\mathrm{def}}{=}
\exp\left(
-\int_0^t
\left[
\delta+
\sum_{\ell=1}^{L}
G_\tau^{*\ell}\mathcal J^\ell(X_\tau)
\right]d\tau
\right),\\
Scf_t
&=
\text{flow i}
+\text{flow ii}
+\text{flow iii},\\
\text{flow i}
&\overset{\mathrm{def}}{=}
\delta
\left(
\frac{\partial\log c}{\partial x}
-
\frac{\partial\hat n}{\partial x}
\right)(X_t),\\
\text{flow ii}
&\overset{\mathrm{def}}{=}
\xi_m
\sum_{\ell=1}^{L}
\mathcal J^\ell(X_t)
\left(
1-\exp\left[
-\frac{V^\ell(X_t)-V(X_t)}{\xi_m}
\right]
\right)
\frac{\partial\log\mathcal J^\ell}{\partial x}(X_t),\\
\text{flow iii}
&\overset{\mathrm{def}}{=}
\sum_{\ell=1}^{L}
g^{*\ell}(X_t)\mathcal J^\ell(X_t)
\frac{\partial V^\ell}{\partial x}(X_t),
\end{align*}
$$

with

$$
G_t^{*\ell}=g^{*\ell}(X_t),
\qquad
g^{*\ell}(X_t)
=
\exp\left[
-\frac{V^\ell(X_t)-V(X_t)}{\xi_m}
\right].
$$
